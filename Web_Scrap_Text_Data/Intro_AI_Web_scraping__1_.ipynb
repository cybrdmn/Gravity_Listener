## What is Web Scraping?

Web scraping is an automated process of extracting publicly available data (data that does not require you to create an account or log in to access it) from the web.

Traditionally, if you want to get information from a website, you would have to manually copy and paste it, which is tedious when dealing with large amounts of data. But a web scraper can obtain data in minutes and export it in your preferred format for further analysis.

As a data scientist, the ability to extract, analyze, and interpret web data is a key skill. You can use web scraping to get everything from sports statistics, product information, stock prices, and much more, which you can use to build datasets. Businesses can also scrape data for market research, brand protection, price and review monitoring, and SEO monitoring.

## Best Practices and Ethical Considerations in Web Scraping

Web scraping raises ethical and legal concerns, even when the data is publicly available. Despite the lack of clear laws or regulations regarding its legality or application, it is crucial to comply with a website’s regulations concerning the data you want to scrape. Here are some best practices to follow before scraping any website:

*   Read the terms of use. Websites have a robots.txt file on the root of their domain that explicitly states if they allow web scraping. Most websites might allow you to scrape their data, as long as it is not for commercial use.
To access the robots.txt file of a website, just add "/robots.txt" to the end of the root domain’s URL. For example, if the root domain is "www.example.com," the robots.txt file will be located at "www.example.com/robots.txt"

*   Set rate limits. Do not spam the website with lots of requests to avoid breaking the website and noticeably reducing the site’s speed for other users.
LearnDataSci
 recommends making only one request per page and parsing only after you have saved the page locally.
*   If the website requires users to log in to access their data, it is wise to reach out to the website and let them know who you are and your intentions.
Do not scrape creative works to avoid breaching laws concerning copyrighted data.





## Basic Understanding of HTML

Websites have different structures even across their subpages and content, so you need to understand their HTML structure to scrape only the data that you want. That’s why I have made crash course on HTML for absolute beginners.

HTML stands for HyperText Markup Language. It is the standard language used to create and design web pages. HTML provides the basic structure and layout for displaying information on the internet. It is the foundation of structuring content of a webpage, including text, images, and other elements.

### Basic Structure
HTML documents are structured using elements enclosed in opening and closing tags. The basic structure of an HTML document begins with a `<!DOCTYPE html>` declaration, `<html>` element, `<head>` element, and `<body>` element.

### Elements and Tags:
HTML consists of various elements and tags that define the structure and content of a webpage. Common elements include headings (`<h1>` to `<h6>`), paragraphs (`<p>`), links (`<a>`), images (`<img>`), lists (`<ul>`, `<ol>`, `<li>`), and more.

### Text Formatting:
HTML allows for text formatting using elements such as `<strong>` for bold text, `<em>` for italicized text, `<u>` for underlined text, and `<br>` for line breaks.

### Links and Images:
Links are created using the `<a>` element with the `href` attribute, while images are inserted using the `<img>` element with the `src` attribute.

### Lists:
HTML supports ordered lists (`<ol>`) and unordered lists (`<ul>`), with list items specified using the `<li>` element.

### Tables:
Tables can be created using the `<table>` element, with rows defined by `<tr>` and cells within the rows specified using `<td>` or `<th>` for header cells.

### Forms:
HTML forms are used to collect user input and are created using the `<form>` element, with input fields such as text fields, checkboxes, radio buttons, and submit buttons.

### Semantic HTML:
Semantic HTML elements like `<header>`, `<footer>`, `<nav>`, and `<section>` provide a clearer structure and meaning to the content, aiding accessibility and search engine optimization.

###Step 2: Import BeautifulSoup and Requests libraries.



```python
from bs4 import BeautifulSoup
import requests
import csv
```

### Step 3: Define the URL of the website you want to scrape, and then send an HTTP GET request to the URL.

```python
url = "https://quotes.toscrape.com/"
# Make sure to include the quotation mark or you will get an error
response = requests.get(url)

# To check if the request was successful (status code 200 means success)
if response.status_code == 200:

    print('Response successful')

else:

    print('Request failed')

#You should get a 'Response Successful'
```

**! Important**
Every request you send to a web server will return a status code, which tells you the outcome of the request and the server’s response to it.
**A status code is like a standardized response that shows if the request was successful or not.** For example, a status code of 200 means everything went well, while 404 means your request couldn’t be found on the server. There are also codes like 400, 401, and 403 that explain specific problems with your request. These codes help you understand what went wrong.

### Best practice: **Make the code robust** (headers, timeouts, errors)

**Important!** requests.get() works on friendly sites but will often fail elsewhere without headers/timeouts/retries.

When scraping websites, we need to communicate with web servers in a careful and respectful way. The code below adds a small layer of robustness and politeness to our requests:

```python
import requests
from requests.exceptions import RequestException, Timeout

BASE_URL = "https://quotes.toscrape.com/"
HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                  "AppleWebKit/537.36 (KHTML, like Gecko) "
                  "Chrome/124.0.0.0 Safari/537.36"
}

def fetch(url, *, timeout=10):
    try:
        resp = requests.get(url, headers=HEADERS, timeout=timeout)
        resp.raise_for_status()
        return resp
    except (RequestException, Timeout) as e:
        print(f"Request failed: {e}")
        return None

response = fetch(BASE_URL)
if response is None:
    raise SystemExit("Stopping: could not fetch the page.")
```

This approach makes our scraper more reliable, respectful, and predictable, especially when scraping real-world websites that may be less friendly than the example site.

Why this is important:

- Websites often block "unknown" or automated scripts.
By sending a User-Agent header, we identify ourselves like a regular browser.
Without this, some websites return 403 Forbidden.

- Things on the internet fail sometimes.
Network delays, temporary outages, redirects, or blocked IPs can happen.
Wrapping the request in a try/except block prevents the program from crashing and allows us to decide what to do next.

- resp.raise_for_status() catches HTTP errors early.
For example, if the page returns 404 or 500, we stop and receive a clear message instead of trying to parse an empty or broken page.

- Timeouts stop the program from hanging indefinitely.
If a website is slow to respond, the script will safely exit instead of freezing.

- Returning None allows the rest of the script to handle failure gracefully.
We can choose to retry, skip, or stop — rather than crash.


### Step 4: Parse the HTML content of the webpage using BeautifulSoup. This shows you the structure of the website and other relevant information.

```python
soup = BeautifulSoup(response.text, "html.parser")
print(soup)
```

### Step 5: Use the `find_all` method to find all the 'span' elements with class 'text' to get the quotes, and all the 'small' elements with class 'author' to get the authors' names.

```python
quotes = soup.find_all('span', attrs={'class': 'text'})

authors = soup.find_all('small', attrs={'class': 'author'})
```

###Step 6: Create an empty list to store the scraped data, then use `for` loop to loop through each pair of quotes and authors, and store them in the variable ‘data’.

```python
data = []


for quote, author in zip(quotes, authors):

    data.append([quote.text.strip(), author.text.strip()])


print(data)
```

###Step 7: Save the file as CSV

```python
csv_filename = 'Quotes_scraped.csv'

with open(csv_filename, 'w', newline='') as file:

    # Creating a CSV writer object

    writer = csv.writer(file)

    # Writing the header row to the CSV file

    writer.writerow(['Quote', 'Author'])

    # Writing the data rows to the CSV file

    writer.writerows(data)
```
